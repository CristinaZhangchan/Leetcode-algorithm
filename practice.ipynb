```python
known = {0:0, 1:1}

def fibonacci_memo(n):
    if n in known:
        return known[n]

    res = fibonacci_memo(n-1)+fibonacci_memo(n-2)
    known[n] = res
    return res
```

```python
def has_duplicate(t):
    seen = set()
    for item in t:
        if item in seen:
            return True
        seen.add(item)
    return False
```

```python
def invert_dict(d):
    new = {}
    for key, value in d.items():
        if value not in new:
            new[value] = [key]
        else:
            new[value].append(key)
    return new
```

```python
def word_distance(word1,word2):
    diff_count = 0
    for char1, char2 in zip(word1, word2):
        if char1 != char2:
            diff_count += 1
    return diff_count
```

```python
def most_frequent_letters(string):
    counter = value_counts(string)
    items = counter.items()
    sorted_items = sorted(items, key=lambda item:item[1], reverse=True)
    for letter, count in sorted_items:
        print(letter, count)
```
